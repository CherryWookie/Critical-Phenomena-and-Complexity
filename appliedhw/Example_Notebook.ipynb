```python
# Definition of the function that generates the BA network

def BA(N, m):
    """
        Inputs:  
                N = size of the network
                m = number of edges emanating from each newly added node

        Returns:
                Network = A dictionary with the adjacency list of the network
    """

    # How to construct the adjacency list

    Network = {}  # empty dictionary
    # fill the keys with the labels of all nodes
    for v in range(N):
        Network[v] = []    # initialize with an empty array

    n0 = m  # size of the initial nucleus
    
    # in the nucleus, every node is connected to every other one
    # initialize the Network dictionary adding one edge for every different pairs of nodes in the nucleus
    
    for v1 in range(n0):
        for v2 in range(n0):
            if v1 != v2:  # no self-connections
                Network[v1].append(v2)
             

    # Fill the body of the function, adding new nodes
    # remeber that the index of the new added nodes ranges from n0 to N-1 !!


    

    return Network
```
